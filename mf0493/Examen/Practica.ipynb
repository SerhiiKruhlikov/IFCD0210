### 1. Una empresa quiere desplegar una aplicación interna solo accesible por sus empleados.
¿Qué tipo de entorno debe utilizar y qué medidas básicas de seguridad implementarías?

### Respuesta
Es necesario utilizar la intranet (red de la empresa), accesible desde la oficina o a través de VPN.

Medidas de seguridad:
* Autenticación: los empleados acceden con sus propias cuentas (integración con LDAP).
* Autorización: diferenciación de permisos (administrador, empleado).
* Cifrado del tráfico: utilizar HTTPS (TLS) dentro de la red.
* Control de acceso por IP: acceso solo desde direcciones IP confiables de la oficina o de la VPN.
* Actualizaciones periódicas: actualización del software en máquinas locales y servidores.
* Registro y monitorización: recopilación de información sobre intentos de acceso y actividades sospechosas.

### 2. Tienes una aplicación web que será accesible desde internet.
Describe los pasos básicos para desplegarla en un servidor Linux.

### Respuesta

* Alquilar un servidor (virtual o dedicado) y configurar el servidor Linux (Ubuntu, Debian, etc.).
* Configurar la red: abrir los puertos 80 y 443 en el cortafuegos.
* Instalar dependencias: servidor web (Nginx/Apache), servidor de aplicaciones (Node.js, Python, PHP), sistema de gestión de bases de datos (PostgreSQL, MySQL) según la tecnología.
* Cargar la aplicación: copiar el código (mediante Git, SFTP u otro medio).
* Configurar el entorno: definir variables (URL de la base de datos, secretos), ajustar permisos de carpetas.
* Configurar el servidor web: redirigir (proxy) las peticiones al servidor de aplicaciones, configurar hosts virtuales.
* Certificado SSL: comprar un certificado (Comodo/Thawte/Symantec) u obtener un certificado gratuito de Let’s Encrypt y configurar HTTPS.
* Iniciar y comprobar: arrancar el servidor de aplicaciones, revisar los logs, abrir el sitio en el navegador.
* Configurar monitorización y copias de seguridad (backups) para detectar fallos a tiempo y poder recuperar los datos.

### 3. Tu aplicación ha sufrido una pérdida de datos crítica.
Explica cómo un sistema de backup y recuperación puede solucionar este problema.

### Respuesta

El sistema de copias de seguridad (backup) permite restaurar los datos a un estado anterior en un momento determinado.

* Las copias de seguridad periódicas copian la base de datos y los archivos de la aplicación a un almacenamiento separado (otro disco, la nube, un servidor independiente).
* Cuando se produce una pérdida de datos, el administrador inicia el proceso de recuperación:
    * Selecciona la marca de tiempo exacta (la última copia de seguridad exitosa anterior al incidente).
    * Restaura los archivos y el volcado de la base de datos en un servidor limpio.
    * Verifica la integridad de los datos.
    * Conmuta la aplicación a los datos recuperados.
* Para minimizar las pérdidas se utiliza la replicación o servidores de respaldo.
* También es importante almacenar las copias de seguridad fuera del sitio principal y probar periódicamente la recuperación.

### 4. Una aplicación necesita almacenar grandes volúmenes de datos.
¿Qué tipo de sistema utilizarías y qué aspectos tendrías en cuenta?

### Respuesta

La elección depende del tipo de datos y de la carga:
* Para datos estructurados y consultas complejas: bases de datos relacionales con escalabilidad, como PostgreSQL, MySQL.
* Para datos no estructurados: sistemas NoSQL, como MongoDB u otros similares.
* Para analítica y big data: almacenes de datos columnar.

Es importante tener en cuenta:
* Escalabilidad: expansión horizontal (añadir nodos).
* Disponibilidad: replicación, tolerancia a fallos.
* Rendimiento: velocidad de lectura/escritura, índices.
* Copia de seguridad: cómo realizar respaldos de los datos.
* Coste: espacio en disco, licencias, tarifas.
* Seguridad: cifrado de datos, control de acceso.

### 5. Tras una actualización, la aplicación deja de funcionar correctamente.
¿Qué proceso seguirías para detectar y solucionar el problema?

### Respuesta

Proceso de detección y resolución del problema (reversión / corrección):

* Detener el deterioro — si la aplicación es crítica, desactivar temporalmente las actualizaciones automáticas o redirigir el tráfico a una versión de respaldo.
* Revisar los registros (logs) — consultar los logs del servidor de aplicaciones, del servidor web y de la base de datos. Normalmente allí aparece el error explícito.
* Reproducir el problema — lo ideal es repetir la acción que causó el fallo en un servidor de pruebas.
* Comparar los cambios — revisar qué archivos, configuraciones o dependencias se actualizaron. Se puede utilizar un sistema de control de versiones (git diff).
* Reversión de la actualización — si el problema es crítico, volver a la versión anterior que funcionaba (restaurar copia de seguridad de archivos y base de datos, retroceder la versión de los paquetes).
* Análisis y corrección — encontrar la causa raíz: incompatibilidad de librerías, error en el código, cambio en la estructura de la base de datos. Preparar el código correcto, actualizar módulos o servicios, eliminar vulnerabilidades en el servidor.
* Nueva actualización — aplicar la versión corregida en el entorno de pruebas, probar y luego en producción (de forma gradual, con monitorización).
* Documentar — registrar la causa y la solución para evitar que se repita.
